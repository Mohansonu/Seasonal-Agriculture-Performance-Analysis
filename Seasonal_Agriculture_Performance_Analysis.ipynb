# **11. Handling Missing Values**

Your dataset has missing values in only 3 columns:

Rainfall_mm → 48 missing values

Soil_Moisture_pct → 40 missing values

Yield_Tonnes_Ha → 32 missing values

**# For now, we will use the median to fill them. Median is a safe and simple method for numerical data.**

**11.1 Fill Missing Rainfall**

```python
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())
```

**11.2 Fill Missing Soil Moisture**

```python
df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(
    df["Soil_Moisture_pct"].median()
)
```

**11.3 Fill Missing Yield**

```python
df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(
    df["Yield_Tonnes_Ha"].median()
)
```

**11.4 Check again**

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

# **2. Profit Margin by Season**

Profit alone does not show efficiency. Profit margin tells us how much of the revenue remains as profit.

```python
df["Profit_Margin_pct"] = (
    df["Profit_INR"] / df["Revenue_INR"]
) * 100
```

```python
profit_margin = df.groupby("Season")["Profit_Margin_pct"].mean()

profit_margin.sort_values(ascending=False)
```

```python
profit_margin.sort_values().plot(
    kind="barh",
    figsize=(8, 5)
)

plt.title("Average Profit Margin by Season")
plt.xlabel("Profit Margin (%)")
plt.ylabel("Season")
plt.show()
```

**What to observe:**

A season can have lower revenue but
still have a higher profit margin
because its production costs may be lower.

# **10. Statistical Conclusion**

ANOVA tests whether the observed differences in average yield across seasons are statistically significant.

```python
from scipy import stats

groups = []

for season in df["Season"].unique():
    groups.append(
        df[df["Season"] == season]["Yield_Tonnes_Ha"]
    )

result = stats.f_oneway(*groups)

print(result)
```

**Interpretation**

If p-value < 0.05:
Seasonal yield differences are statistically significant.

If p-value >= 0.05:
There is insufficient statistical evidence to conclude
that yield differs significantly across seasons.

**Important:**
Statistical significance does not prove that the season itself caused the difference.
